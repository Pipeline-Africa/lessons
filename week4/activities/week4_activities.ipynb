## 1. Group Discussion

Spend 10–12 minutes per topic.

### Discussion Topic 1 — Flat Files vs Databases: When Does It Matter?

This week you saved data to CSV and then loaded it into PostgreSQL. Both can hold the same data — so why bother with the database step at all?

**Opening question:** Your pipeline runs every day and adds 5 rows of weather data. After one year you have 1,825 rows. After five years you have 9,125 rows. At what point, if any, does loading into a CSV feel like the wrong choice?

**Dig deeper:**
- What happens if two scripts try to write to the same CSV at the same time?
- What if you need to query the last 30 days of data for one city only? How do you do that in a CSV? How do you do it in SQL?
- What did the upsert pattern solve that a plain CSV cannot?
- Can you think of a use case where CSV is the *better* choice even when a database is available?

**Share:** Each group gives one real-world scenario where they would use CSV and one where they would use a database.

### Discussion Topic 2 — What Can Go Wrong in a Pipeline That Calls an API?

A pipeline that calls an API has more failure points than one that reads local files.

**Opening question:** Your weather pipeline runs at 6 AM every day. You wake up at 9 AM and see it produced no output. What is your list of things to check?

**Dig deeper:**
- The API returns a `200 OK` status but the JSON body is missing the `daily` key. Your pipeline crashes. How would you catch this?
- The API starts returning data with an extra column you were not expecting. What happens to your pipeline?
- You are charged per API call. Your pipeline has a bug and calls the API 10,000 times in one minute. What safeguards could prevent this?
- How is a log file useful in each of these failure scenarios?

**Share:** Each group names the one safeguard they would add to a production API pipeline first.

### Discussion Topic 3 — Secrets, Config, and the Right Way to Handle Credentials

**Opening question:** A new team member joins and needs to run the weather pipeline. You send them the code. What should they need to change to make it work on their machine? What should they *not* need to change?

**Dig deeper:**
- What happens if someone commits a `.env` file with a real password to GitHub? (This happens more often than you think.)
- A `.gitignore` file prevents files from being committed. Where should `.env` appear in a `.gitignore`?
- You have three environments: local development, staging, and production. Each has a different database. How would you structure your config and environment variables to support this?
- Why is it safer to store a secret in an environment variable than in a JSON config file?

**Share:** Each group writes a one-line rule about secrets that everyone on the team should follow.
