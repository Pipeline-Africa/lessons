## 1. Group Discussion

Spend 10–12 minutes per topic.

### Discussion Topic 1 — Is This Scraper Allowed?

This week you took data from four websites. None of them asked you first.

**Opening question:** Your manager asks you to scrape competitor prices from five retail websites every morning and load them into a dashboard. What do you check before you write a line of code — and what would make you say no?

**Dig deeper:**
- `robots.txt` is a text file with no enforcement behind it. If nothing stops your code from ignoring it, why does it matter?
- A site's Terms of Service prohibit automated access. Does that make scraping it illegal, or just a breach of contract? Does your answer change if you had to click "I agree" to see the page?
- One of the five sites shows prices only to logged-in users. Your manager offers you an account. Where is the line, and which side of it is that on?
- You scrape a jobs site and the pages include the recruiter's name, email and phone number. What does Nigeria's NDPR (or the GDPR) have to say about your pipeline now? What would you do differently?
- Your scraper, left running with a bug, sends 50,000 requests an hour to a small business's website. What is the difference between that and a denial-of-service attack — intent, or effect?

**Share:** each group writes the three checks they would run *before* starting any scraping job, in order.

### Discussion Topic 2 — Why Is Scraped Data Harder Than API Data?

Last week you got JSON with types, field names and a documented contract. This week you got whatever the HTML happened to contain.

**Opening question:** You have two pipelines in production: one calls a REST API, one scrapes a job board. Both run at 6 AM. Which one wakes you up at night more often, and why?

**Dig deeper:**
- The site redesigns and `.job-card` becomes `.posting-card`. Your pipeline returns 0 rows but does not crash. How would you have detected this on the first morning instead of the third week?
- What would you assert about a scrape before loading it into the warehouse? Row count? Null rate? Something else?
- Lab 5 revealed that every posting on the board had the same date. Your validation step did not fail — you spotted it by looking. How do you build the habit of looking, and what *can* be automated?
- An API gives you a documented contract. What is the equivalent for a scraper — is there one?
- When a site both offers an API and shows the data on a page, is there ever a reason to scrape the page anyway?

**Share:** each group proposes one automated check that would have caught a silent scraper failure, and says what it would cost in false alarms.

### Discussion Topic 3 — Formats, Types and the Cost of CSV

In Lesson 4 the same 20,000 rows ranged from under a megabyte to several megabytes depending only on the format.

**Opening question:** Your team's pipeline passes CSV files between every stage and has done for two years. Nothing is on fire. Make the case for changing it — and then make the case against.

**Dig deeper:**
- Which is the stronger argument for Parquet: the smaller files, or the preserved types? Which one will your manager care about, and which one will cost you a weekend if you ignore it?
- A CSV read with the wrong inferred dtype does not raise an error — it produces slightly wrong numbers. Where in a pipeline would that do the most damage before anyone noticed?
- Parquet files cannot be opened in a text editor or in Excel. What do you lose operationally when your data stops being human-readable, and how do you compensate?
- Your partitioned dataset came out 100× larger until you sorted it first. What does that tell you about trusting a "best practice" you have not measured?
- When is CSV genuinely the right answer, and not just the lazy one?

**Share:** each group writes one rule about file formats that they would put in their team's engineering handbook.
